```python
# Experiment for applying AC pulse using Cluster during DC sweep using SPI Rack through SA. Acquisition through DMM and SA.
# Give the Frequency, Total Duration of the pulse, and the number of repeats for a on/off cycle.

Frequency = 5000 # Hz
duration = 220e9 # ns

qcm_output_name = ["O1"]
pulse_amplitude = [0.5]
pulse_length = [1e9/Frequency]
times_between_pulses = [1e9/Frequency]
acquisition_time = [0]
acquisition_delay = [0.0]
acquisition_resolution = [1]
repeats = [int(duration/(pulse_length[0]+times_between_pulses[0]))]

new_experiment.run_square_pulse(qcm_output_name,
						pulse_amplitude,
						pulse_length,
						times_between_pulses,
						acquisition_time,
						acquisition_delay,
						acquisition_resolution,
						repeats,    
						voltage_configuration = {},
						plot = False)
```

```python
# Experiment for constant current measurement.

start_amp = 0.0 # V
stop_amp = 1.0 # V
num_points = 101 # remember to add 1 to include the stop_amp in the sweep
step_size = (stop_amp - start_amp)/(num_points-1)
Frequency = 5000 # Hz
averages = 100
qcm_output_name = ["O1"]
pulse_length = [1e9/Frequency]
times_between_pulses = [100]
acquisition_time = [0]
acquisition_delay = [0.0]
acquisition_resolution = [1]
repeats = [1]

sweep_amps = np.arange(start_amp, stop_amp + step_size, step_size)

for i in sweep_amps:
    pulse_amplitude = [i]
    for j in range(averages):
        new_experiment.run_square_pulse(qcm_output_name,
                                        pulse_amplitude,
                                        pulse_length,
                                        times_between_pulses,
                                        acquisition_time,
                                        acquisition_delay,
                                        acquisition_resolution,
                                        repeats,    
                                        voltage_configuration = {},
                                        plot = False)
```
